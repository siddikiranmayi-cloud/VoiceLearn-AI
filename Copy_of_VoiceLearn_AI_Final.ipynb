```python
from IPython.display import display, Javascript
from google.colab.output import eval_js
from base64 import b64decode

def record_audio(seconds=8):
    display(Javascript("""
    async function recordAudio(seconds) {
        const stream = await navigator.mediaDevices.getUserMedia({audio: true});
        const recorder = new MediaRecorder(stream);
        const chunks = [];

        recorder.ondataavailable = e => chunks.push(e.data);
        recorder.start();

        await new Promise(resolve => setTimeout(resolve, seconds * 1000));
        recorder.stop();

        await new Promise(resolve => recorder.onstop = resolve);
        stream.getTracks().forEach(track => track.stop());

        const blob = new Blob(chunks, {type: 'audio/webm'});
        const reader = new FileReader();

        return await new Promise(resolve => {
            reader.onloadend = () => resolve(reader.result);
            reader.readAsDataURL(blob);
        });
    }
    """))

    audio_data = eval_js(f"recordAudio({seconds})")
    audio_bytes = b64decode(audio_data.split(",")[1])

    with open("VoiceLearn_Question.webm", "wb") as f:
        f.write(audio_bytes)

    print("Recording saved successfully!")
    return "VoiceLearn_Question.webm"

audio_file = record_audio(8)
```

```python
import requests
import os
from IPython.display import Audio, display

url = "https://api.vachana.ai/api/v1/tts/inference"

headers = {
    "X-API-Key-ID": os.environ["GNANI_API_KEY"],
    "Content-Type": "application/json"
}

payload = {
    "text": study_answer[:700],
    "voice": "Kaveri",
    "model": "timbre-v2.5",
    "language": "en-IN",
    "speed": 1.0,
    "audio_config": {
        "encoding": "linear_pcm",
        "container": "wav",
        "num_channels": 1,
        "sample_rate": 48000,
        "sample_width": 2
    }
}

try:
    tts_response = requests.post(
        url,
        headers=headers,
        json=payload,
        timeout=90
    )

    print("Timbre status:", tts_response.status_code)
    tts_response.raise_for_status()

    with open("VoiceLearn_Final_Audio.wav", "wb") as f:
        f.write(tts_response.content)

    print("Voice answer generated successfully!")
    display(Audio("VoiceLearn_Final_Audio.wav"))

except Exception as e:
    print("Timbre error:", e)
    if 'tts_response' in globals():
        print(tts_response.text[:500])
```
